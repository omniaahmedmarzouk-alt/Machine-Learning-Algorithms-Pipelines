```python
import random
```

```python
def simple_trick( bias, weight, X, Y, learning_rate):
    predicted_y= bias + (weight * X)
    if Y > predicted_y:
        if X > 0:
            weight += learning_rate
            bias += learning_rate
        else:
            weight -= learning_rate
            bias += learning_rate
    elif Y < predicted_y:
        if X > 0:
            weight -= learning_rate
            bias -= learning_rate
        else:
            weight += learning_rate
            bias -= learning_rate
    return weight, bias
```

```python
def absolute_trick(bias, weight, X, Y, learning_rate):
    predicted_y = bias + (weight * X)
    if Y > predicted_y:
        weight += learning_rate * X
        bias += learning_rate
    elif Y < predicted_y:
        weight -= learning_rate * X
        bias -= learning_rate
    return weight, bias
```

```python
def square_trick( bias, weight, X, Y, learning_rate):
   predicted_y= bias + (weight * X)
   bias+= learning_rate *( Y - predicted_y)
   weight+= X * learning_rate *( Y - predicted_y)

   return bias, weight
```

```python
def Linear_regression( X, Y, epochs=1000, learning_rate=0.01, trick_name='square'):
  weight= random.random()
  bias= random.random()
  for epoch in range(epochs):
     i= random.randint(0,len(X)-1)
     x=X[i]
     y=Y[i]
     if trick_name == "simple":
            weight, bias = simple_trick(bias, weight, x, y, learning_rate)
     elif trick_name == "absolute":
            weight, bias = absolute_trick(bias, weight, x, y, learning_rate)
     elif trick_name == "square":
            weight, bias = square_trick(bias, weight, x, y, learning_rate)
  return bias, weight
```

```python

features = [1.0, 2.0, 3.0, 4.0]
labels = [2.0, 4.0, 6.0, 8.0]
epochs = 1000
learning_rate = 0.01

#Simple Trick
w_simple, b_simple = Linear_regression(features, labels, epochs, learning_rate, trick_name="simple")
print(f"Simple Trick -> Weight: {w_simple:.2f}, Bias: {b_simple:.2f}")

#Absolute Trick
w_abs, b_abs = Linear_regression(features, labels, epochs, learning_rate, trick_name="absolute")
print(f"Absolute Trick -> Weight: {w_abs:.2f}, Bias: {b_abs:.2f}")

#Square Trick
w_square, b_square =Linear_regression(features, labels, epochs, learning_rate, trick_name="square")
print(f"Square Trick -> Weight: {w_square:.2f}, Bias: {b_square:.2f}")
```
