```python
import numpy as np

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 4, 5])

m = 0 # (Slope)
b = 0 # (Intercept)
L = 0.01 # Learning Rate 
epochs = 1000 

# Gradient Descent Loop
for i in range(epochs): 
    y_pred = m*X + b 
    # (Derivatives)
    D_m = (-2/len(X)) * sum(X * (y - y_pred))
    D_b = (-2/len(X)) * sum(y - y_pred)
    # (Updating Parameters)
    m = m - L * D_m
    b = b - L * D_b

print(f"Bset Slope value: {m}")
print(f"Best Intercept value: {b}")
```
